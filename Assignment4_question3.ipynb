**Question 3 — CNOT from H and CZ only**  

Build an exact copy of CNOT using only H and CZ gates, no cx allowed, and verify the matrix matches.

```python
%pip install qiskit
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, Operator
from qiskit.visualization import plot_histogram
```

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Operator

# 1. Build equivalent CNOT using H and CZ
qc_equivalent = QuantumCircuit(2)
qc_equivalent.h(1)       # Hadamard on target (q1)
qc_equivalent.cz(0, 1)   # CZ with control=q0, target=q1
qc_equivalent.h(1)       # Hadamard on target (q1)
print("qc_equivalent: ",qc_equivalent)
state_qc_equivalent = Statevector(qc_equivalent)
print("qc equivalent Statevector: ", state_qc_equivalent)
#
# 2. Build standard CNOT
qc_cnot = QuantumCircuit(2)
qc_cnot.cx(0, 1)         # Standard CNOT with control=q0, target=q1
print("qc_cnot: ", qc_cnot)
state_qc_cnot = Statevector(qc_cnot)
print("standard cnot Statevector: ", state_qc_cnot)

# 3. Extract matrix operators
op_equivalent = Operator(qc_equivalent)
op_cnot = Operator(qc_cnot)

# 4. Confirm equality
print("Matrices match:", op_equivalent.equiv(op_cnot))
```

The above circuit qc_equivqlent acts identically to a standard CNOT for any input state.
1. Applying Hardamard gate on q1:

          Converts q1 from the computational basis (|0>, |1>) to the diagonal basis (|+>, |->).
      
          Applying CZ Gate:
            If q0 = 0,
            The CZ gate does nothing to q1.
            If q0 = 1,
            The CZ gate applies a Z phase shift to q1.
            In the diagonal basis,
            a Z gate swaps |+> --> |->.

2. Applying Hardamard Gate on q1:
    
          Converts q1 back to the computational basis.
          
          If q1 was unchanged, H.H = I, leaving it as it was.
          If q1 was flipped in the diagonal basis |+> --> |->,
          the second H gate converts that into a standard computational flip |0> --> |1>.
          Because HZH = X, placing H gates around a CZ gate operation transforms the target action from Z to X, creating a Controlled-X(CNOT) gate.
