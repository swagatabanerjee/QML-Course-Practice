**Question 6 — SWAP from CNOTs only**  


Build an exact copy of SWAP using only CNOT gates, no swap allowed, and verify the matrix matches.

```python
%pip install qiskit
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, Operator
from qiskit.visualization import plot_histogram
```

```python
# 1. Build SWAP circuit using only 3 CNOT gates
qc_cnot_swap = QuantumCircuit(2)
qc_cnot_swap.cx(0, 1)  # CNOT 1
qc_cnot_swap.cx(1, 0)  # CNOT 2
qc_cnot_swap.cx(0, 1)  # CNOT 3
print("qc_cnot_swap: ",qc_cnot_swap)
# 2. Build standard SWAP circuit
qc_swap = QuantumCircuit(2)
qc_swap.swap(0, 1)
print("qc_swap: ", qc_swap)
# 3. Extract and verify matrix equality
op_cnot_swap = Operator(qc_cnot_swap)
op_swap = Operator(qc_swap)

print("Matrices match:", op_cnot_swap.equiv(op_swap))  # Outputs: True
```
