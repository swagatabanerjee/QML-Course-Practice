**Question 4 — CNOT on an already-1 control**

Predict the result of applying CNOT to the state |11⟩ before building it, then confirm your prediction.

```python
%pip install qiskit
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, Operator
from qiskit.visualization import plot_histogram
```

**Initial State:**
      
      |q1 q0>= |11> (q0 = 1, q1 = 1)
  Apply CNOT Gate(qc.cx(0, 1) )
   
    Control qubit q0=1
    Target qubit q1=1
    Because the control qubit q0 is in state |1>,
    the target qubit q1 flips from 1 --> 0.
    
    The control qubit q0 remains unchanged (1).
  
  **Final State:** |01>
  
  When measured, the circuit will yield the outcome 01 with expected $100 probability.

```python
# Initialize a 2-qubit circuit
qc = QuantumCircuit(2)

# Step 1: Prepare the initial state |11>
qc.x(0)  # q0 = 1
qc.x(1)  # q1 = 1

# Step 2: Apply CNOT (q0 = control, q1 = target)
qc.cx(0, 1)

# Step 3: Verify the statevector
state = Statevector.from_instruction(qc)
print(qc)
plot_histogram(state.probabilities_dict()) # Outputs |01>
```
